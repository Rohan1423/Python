```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('tested_fare_imputed.csv',usecols=['Age','Pclass','Fare','Survived'])
df.head()
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
X_train.head()
```

```python
knn = KNNImputer(n_neighbors=3,weights='distance') # by default n_neighbors = 5 and weights = uniform
# knn = SimpleImputer() # giving 64 % (less than KNNImputer)
X_train_trf = knn.fit_transform(X_train)
X_test_trf = knn.transform(X_test)
```

```python
pd.DataFrame(X_train_trf,columns=X_train.columns)
```

```python
lr = LogisticRegression()
lr.fit(X_train_trf,y_train)

y_pred = lr.predict(X_test_trf)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
