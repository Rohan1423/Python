```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('covid_toy.csv')
df.head()
df['cough'].value_counts()
df.isnull().sum()
```

```python
# Here Gender and City are Nominal Categorical, Cough is Ordinal Categorical and Has_Covid(y) is Nominal Categorical. Also Fever is Numerical Column but it has Values missing in it

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop('has_covid',axis=1),df['has_covid'],test_size=0.2,random_state=0)
X_train
```

# Encoding Without using Column Transformer

```python
# Applying SI to Fever column for Missing Values -> Mean Value
from sklearn.impute import SimpleImputer
si = SimpleImputer()

X_train_fever = si.fit_transform(X_train[['fever']])
X_test_fever = si.transform(X_test[['fever']])

X_train_fever
```

```python
# Applying OE to Ordinal Data
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['Mild','Strong']])

X_train_cough = oe.fit_transform(X_train[['cough']])
X_test_cough = oe.transform(X_test[['cough']])

X_train_fever
```

```python
# Applying OHE to Nominal Data
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop='first',sparse_output=False)

X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city = ohe.transform(X_test[['gender','city']])

X_train_gender_city
```

```python
# Extracting Age and convert it to NumPy Array So we can combine it with Other Numpy arrays we created
X_train_age = X_train[['age']].to_numpy()
X_test_age = X_test[['age']].to_numpy()
```

```python
# Combining all Numpy arrays
X_train_transformed = np.concatenate((X_train_age,X_train_fever,X_train_gender_city,X_train_cough),axis=1)
X_test_transformed = np.concatenate((X_test,X_test_fever,X_test_gender_city,X_test_cough),axis=1)
X_train_transformed
```

# Encoding With using Column Transformer

```python
from sklearn.compose import ColumnTransformer
# transformers=[] : Define all column-wise transformations here.
# Syntax for each transformer:
# ('transformer_name', transformer_object(), ['columns_to_apply'])

# remainder : Controls the columns that are not mentioned in 'transformers'.
# 'drop'        -> Ignore those columns.
# 'passthrough' -> Include those columns in the final output without any transformation.

# Create a ColumnTransformer to apply different transformation techniques to different columns in a single step.
transformer = ColumnTransformer(
    transformers=[
        # tnf1: Replace missing values in the 'fever' column using SimpleImputer
        ('tnf1', SimpleImputer(), ['fever']),

        # tnf2: Apply Ordinal Encoding to the 'cough' column
        # Mapping: Mild -> 0, Strong -> 1
        ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),

        # tnf3: Apply One-Hot Encoding to 'gender' and 'city'
        # drop='first' removes the first category to avoid the dummy variable trap
        # sparse_output=False returns a NumPy array instead of a sparse matrix
        ('tnf3', OneHotEncoder(drop='first', sparse_output=False), ['gender', 'city'])
    ],

    # Keep all remaining columns (Age) unchanged
    remainder='passthrough'
)

# Learn all preprocessing parameters from the training data
X_train_transformed = transformer.fit_transform(X_train)

# Apply the same preprocessing to the test data
X_test_transformed = transformer.transform(X_test)

X_train_transformed
```
