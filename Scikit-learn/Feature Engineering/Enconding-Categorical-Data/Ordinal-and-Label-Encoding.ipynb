```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('customer.csv')
df.head(10)
# Gender is nominal categorical, Review is Ordinal Categorical, Education is Ordinal Categorical, Purchased is Nominal Categorical
```

```python
# Here we will apply Nominal Encoding to gender, Ordinal Encoding to review & education, Label Encoding to purchased
df = df.iloc[:,2:]
df.head(10)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop('purchased',axis=1),df['purchased'],test_size=0.2,random_state=0)
print(X_train.columns)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']]) # here the order is 0,1,2 for the Ordinal categorical values
oe.fit(X_train)
X_train = oe.transform(X_train)
X_test = oe.transform(X_test)
```

```python
oe.categories_
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
le.fit(y_train)
y_train = le.transform(y_train)
y_test = le.transform(y_test)
```
