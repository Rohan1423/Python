```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('placement.csv',usecols=['cgpa','package'])
df.head()
```

```python
# plt.scatter(x,y)
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package(inLPA)')
```

```python
from sklearn.model_selection import train_test_split

X = df.iloc[:,0:1]
y = df.iloc[:,1]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
X_train
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
# Here we are predicting a particular value of X_test we are reshaping beacuse it is a single value so converting it to 1x1 and doing .values(convert to numpy) so that we can do .reshape
y_pred = lr.predict(X_test)
```

```python
X_test
```

```python
y_test
```

```python
# plt.scatter(x,y)
plt.scatter(df['cgpa'], df['package'])

# Linear Regression Best Fit Line
plt.plot(
    X_train,
    lr.predict(X_train),
    color='red',
    linestyle='-',   # Dashed line
    linewidth=3
)

plt.xlabel('CGPA')
plt.ylabel('Package (in LPA)')
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
print("MAE",mean_absolute_error(y_test,y_pred))
```

```python
print("MSE",mean_squared_error(y_test,y_pred))
```

```python
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
print("R2 Score",r2_score(y_test,y_pred))
```

```python
r2 = r2_score(y_test,y_pred)
X_test.shape
```

```python
print("Adjusted R2 Score",1 - ((1-r2)*(40-1))/(40-1-1))
```
