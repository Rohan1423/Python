```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('placement.csv',usecols=['cgpa','package'])
df.head()
```

```python
# plt.scatter(x,y)
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package(inLPA)')
```

```python
from sklearn.model_selection import train_test_split

X = df.iloc[:,0:1]
y = df.iloc[:,1]
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
X_train
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
# Here we are predicting a particular value of X_test we are reshaping beacuse it is a single value so converting it to 1x1 and doing .values(convert to numpy) so that we can do .reshape
lr.predict(X_test.iloc[0].values.reshape(1,1))
```

```python
X_test
```

```python
y_test
```

```python
# plt.scatter(x,y)
plt.scatter(df['cgpa'], df['package'])

# Linear Regression Best Fit Line
plt.plot(
    X_train,
    lr.predict(X_train),
    color='red',
    linestyle='-',   # Dashed line
    linewidth=3
)

plt.xlabel('CGPA')
plt.ylabel('Package (in LPA)')
plt.show()
```

```python
# Slope(m)
m = lr.coef_
m
```

```python
# Intercept(b)
b = lr.intercept_
b
```

```python
# Example of Calculating the value of X_test[0] cgpa value using this equation 
y = m * 8.58 + b
# Getting same value which our model predicted
y
```
